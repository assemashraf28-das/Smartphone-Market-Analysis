```python
def extract_brand(title):
    
    title = str(title).lower()

    if 'samsung' in title:
        return 'Samsung'
    elif 'iphone' in title or 'apple' in title:
        return 'Apple'
    elif 'motorola' in title or 'moto' in title:
        return 'Motorola'
    elif 'google' in title or 'pixel' in title:
        return 'Google'
    elif 'oneplus' in title:
        return 'OnePlus'
    elif 'nokia' in title:
        return 'Nokia'
    elif 'blu' in title:
        return 'BLU'
    elif 'tcl' in title:
        return 'TCL'
    elif 'alcatel' in title:
        return 'Alcatel'
    else:
        return 'Other'
```

```python
import re

def extract_storage(title):

    title = str(title)

    match = re.search(r'(\d+)\s?GB', title, re.IGNORECASE)

    if match:
        return int(match.group(1))

    return np.nan
```
